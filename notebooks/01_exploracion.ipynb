## Observaciones sobre la columna `Price`

La columna `Price` contiene información de dos variables: la moneda y el importe.

Se identificaron publicaciones en pesos argentinos (`$`) y en dólares estadounidenses (`USD`).

Aproximadamente el 60% de las publicaciones están expresadas en pesos y el 40% en dólares.

Será necesario separar la moneda del valor numérico para facilitar el análisis.

```python
# Extrae la moneda de Price y la guarda en una nueva columna llamada "currecy"
df_clean["currency"] = df_clean["Price"].str.extract(r"(USD|\$)")[0]
```

```python
# Comprueba cuantos registros hay en cada moneda
df_clean["currency"].value_counts()
```

## Limpieza de la columna `Price`

La columna `Price` contiene dos tipos de información:

- La moneda (`USD` o `$`).
- El valor del alquiler.

Se extrae el importe numérico y se almacena en una nueva columna llamada `price_value`, manteniendo la columna original `Price` para conservar el dato tal como fue obtenido.

```python
# Elimina separadores y símbolos de moneda 
# para preparar el precio para convertirlo a número.
price_test = pd.to_numeric(
    df_clean["Price"]
    .str.replace(".", "", regex=False)
    .str.replace("USD", "", regex=False)
    .str.replace("$", "", regex=False),
    errors="coerce"
)

# Cuenta los valores que no pudieron convertirse a número
price_test.isna().sum()
```

```python
# Identifica qué valores originales provocaron conversiones inválidas.
df_clean.loc[price_test.isna(), "Price"].value_counts()
```

```python
# Guarda el precio numérico limpio en price_value.
df_clean["price_value"] = price_test
```

```python
# Verifica visualmente la correspondencia entre 
# precio original y precio numérico.
df_clean[["Price", "price_value"]].head(10)
```

```python
# Comprueba cuántos precios quedaron como valores faltantes.
df_clean["price_value"].isna().sum()
```

## Revisión de precios elevados en dólares

Los precios expresados en USD requieren un análisis separado, ya que un valor elevado puede corresponder a una publicación incorrecta, a una modalidad de alquiler diferente o a información adicional incluida en la descripción.

Se revisan especialmente los registros con precios iguales o superiores a USD 10.000.

```python
# Obtiene los alquileres en USD con valores 
# muy elevados para inspeccionarlos.
usd_altos = df_clean[
    (df_clean["currency"] == "USD") &
    (df_clean["price_value"] >= 10_000)
].copy()
print("Registros USD >= 10.000:", len(usd_altos))

     
```

## Revisión de valores USD extremadamente altos

Se realiza una segunda revisión sobre precios iguales o superiores a USD 50.000 para determinar si corresponden realmente al alquiler mensual o si representan otra modalidad de publicación.

```python
# Aísla los casos con precios USD extremadamente altos.
usd_muy_altos = df_clean.loc[
    (df_clean["currency"] == "USD") &
    (df_clean["price_value"] >= 50_000),
    ["id", "Price", "price_value", "Location", "Features", "Description"]
]

print("Registros USD >= 50.000:", len(usd_muy_altos))

print(
    usd_muy_altos[
        ["id", "Price", "price_value", "Location", "Features"]
    ].to_string()
)
```

## Identificación de alquileres temporarios

Los alquileres temporarios pueden presentar precios expresados por día, semana, noche o estadía, por lo que sus valores no son directamente comparables con los alquileres mensuales.

Se crea una variable que permite identificarlos a partir de términos presentes en la descripción.

```python
# Marca publicaciones que mencionan modalidades de alquiler temporario.
df_clean["es_temporario"] = df_clean["Description"].str.contains(
    r"alquiler temporario|alquiler temporal|temporada|por día|por dias|diario|por semana|semanal|quincena|quincenal|estadía|estadia|mínimo.*días|minimo.*dias|noches|por noche",
    case=False,
    na=False,
    regex=True
)

print("Registros temporarios:", df_clean["es_temporario"].sum())
```

```python
# Revisa los precios menores a USD 100 que fueron 
# identificados como temporarios.
print(
    df_clean.loc[
        (df_clean["price_value"] < 100) &
        (df_clean["es_temporario"]) &
        (df_clean["currency"] == "USD"),
        ["id", "Price", "price_value", "currency", "Description"]
    ].head(10)
)
```

```python
#Busca precios USD menores a 100 que no fueron clasificados
# como temporarios.
print(
    df_clean.loc[
        (df_clean["price_value"] < 100) &
        (df_clean["currency"] == "USD") &
        (~df_clean["es_temporario"]),
        ["id", "Price", "price_value", "Description"]
    ].head(20).to_string()
)
```

## Identificación de referencias a alquileres anuales

Algunas publicaciones pueden mostrar en el precio un importe correspondiente al primer año, segundo año o al valor anual.

Se crea una variable para identificar estas referencias en la descripción y evitar interpretar automáticamente estos importes como alquileres mensuales.

```python
# Detecta descripciones que mencionan valores anuales o por año.
df_clean["es_anual"] = df_clean["Description"].str.contains(
    r"alquiler anual|alquileres anual|valor primer año|primer año.*usd|segundo año",
    case=False,
    na=False,
    regex=True
)

print("Registros posiblemente anuales:", df_clean["es_anual"].sum())
```

```python
# Muestra ejemplos de publicaciones clasificadas como 
# posiblemente anuales.
print(
    df_clean.loc[
        df_clean["es_anual"],
        ["id", "Price", "price_value", "currency", "Description"]
    ].head(20)
)
```

## Revisión de precios USD menores a 100

Los valores inferiores a USD 100 pueden corresponder a alquileres temporarios, importes parciales, referencias dentro de la descripción u otros casos particulares.

Por este motivo, no se eliminan automáticamente y se revisan según el contexto de cada publicación.

```python
# Identifica precios USD menores a 100 para su revisión.
df_clean["precio_usd_sospechoso"] = (
    (df_clean["currency"] == "USD") &
    (df_clean["price_value"] < 100)
)

print(
    "Precios USD menores a 100:",
    df_clean["precio_usd_sospechoso"].sum()
)
```

```python
# Muestra algunos casos sospechosos junto con su contexto.
df_clean.loc[
    df_clean["precio_usd_sospechoso"],
    ["id", "Price", "price_value", "es_anual", "es_temporario", "Description"]
].head(20)
```

## Identificación de posibles publicaciones de venta

Algunas publicaciones presentan precios elevados en dólares y contienen referencias explícitas a operaciones de venta.

Se crea una variable auxiliar para identificar estas publicaciones y evitar que puedan distorsionar el análisis de alquileres.

```python
# Detecta menciones de venta dentro de las descripciones.
df_clean["menciona_venta"] = df_clean["Description"].str.contains(
    r"\bventa\b|en venta|precio venta|venta directa|venta urgente|venta con renta",
    case=False,
    na=False,
    regex=True
)

print("Registros que mencionan venta:", df_clean["menciona_venta"].sum())
```

```python
# Identifica los precios USD elevados que además mencionan venta.
usd_altos_venta = df_clean.loc[
    (df_clean["currency"] == "USD") &
    (df_clean["price_value"] >= 10_000) &
    (df_clean["menciona_venta"]),
    ["id", "Price", "price_value", "Location", "Features", "Description"]
]

print(
    "USD >= 10.000 que mencionan venta:",
    len(usd_altos_venta)
)
```

```python
# Resume cómo se clasifican los precios USD elevados según las 
# variables detectadas.
print(
    df_clean.loc[
        (df_clean["currency"] == "USD") &
        (df_clean["price_value"] >= 10_000),
        ["es_anual", "es_temporario", "menciona_venta"]
    ].value_counts()
)
```

## Clasificación de posibles publicaciones de venta

Se define una condición para identificar publicaciones que cumplen simultáneamente con tres características:

- Precio expresado en USD.
- Precio igual o superior a USD 10.000.
- Mención de venta en la descripción.

Estas publicaciones se consideran posibles casos de venta y se mantienen identificadas para decidir posteriormente cómo tratarlas en el análisis.

```python
# Marca los registros que cumplen las condiciones 
# definidas como posibles ventas.

df_clean["posible_venta"] = (
    df_clean["currency"].eq("USD") &
    (df_clean["price_value"] >= 10_000) &
    df_clean["menciona_venta"]
)

print(
    "Posibles publicaciones de venta:",
    df_clean["posible_venta"].sum()
)
```

```python
# Revisa ejemplos de publicaciones clasificadas como posibles ventas.
print(
    df_clean.loc[
        df_clean["posible_venta"],
        ["id", "Price", "price_value", "Location", "Features"]
    ].head(20).to_string()
)
```

```python
# Comprueba la cantidad de publicaciones y propiedades 
# únicas detectadas.
print(
    "Registros posibles venta:",
    df_clean["posible_venta"].sum()
)

print(
    "IDs únicos:",
    df_clean.loc[
        df_clean["posible_venta"],
        "id"
    ].nunique()
)
```

## Revisión de posibles ventas repetidas

Un mismo inmueble puede aparecer en distintas fechas de scraping. Por lo tanto, se analiza la frecuencia de los `id` identificados como posibles ventas para distinguir entre publicaciones repetidas y propiedades diferentes.

```python
# Cuenta cuántas veces aparece cada propiedad entre 
# las posibles ventas.
duplicados_venta = (
    df_clean.loc[df_clean["posible_venta"], "id"]
    .value_counts()
)

print(duplicados_venta)
```

```python
# Obtiene todos los IDs correspondientes a posibles ventas.
ids_venta = df_clean.loc[
    df_clean["posible_venta"],
    "id"
].unique()
```

```python
# Comprueba cuántas veces aparece cada propiedad en todo el dataset.
print(
    df_clean[
        df_clean["id"].isin(ids_venta)
    ]["id"].value_counts()
)
```

```python
# Compara las apariciones de cada propiedad a través de 
# las fechas de scraping.
print(
    df_clean[
        df_clean["id"].isin(ids_venta)
    ][
        ["id", "Price", "Location", "scrap_date"]
    ].sort_values("id").to_string()
)
```

## Análisis de cambios de precio

Como una misma propiedad puede aparecer en diferentes fechas de scraping, se analiza si el precio registrado para un mismo `id` cambia con el tiempo.

Esto permite diferenciar entre duplicados producto del proceso de scraping y modificaciones reales en la publicación.

```python
# Cuenta cuántos precios distintos tuvo cada propiedad.
cambios_precio = (
    df_clean
    .groupby("id")["price_value"]
    .nunique()
)

print(
    "IDs con más de un precio:",
    (cambios_precio > 1).sum()
)
```

```python
# Muestra las propiedades con mayor cantidad de precios diferentes.
print(
    cambios_precio[cambios_precio > 1]
    .sort_values(ascending=False)
    .head(20)
)
```

```python
# Selecciona las propiedades cuyos precios cambiaron para revisar su 
# evolución temporal.
ids_cambio_precio = cambios_precio[cambios_precio > 1].index

cambios = (
    df_clean[df_clean["id"].isin(ids_cambio_precio)]
    [["id", "Price", "price_value", "currency", "Location", "scrap_date"]]
    .sort_values(["id", "scrap_date"])
)
```

```python
# Muestra ejemplos de cambios de precio ordenados cronológicamente.

print(
    cambios.head(50).to_string(index=False)
)
```

## Análisis de cambios de moneda

Se analiza si una misma propiedad aparece registrada en diferentes monedas a lo largo de las fechas de scraping.

Estos casos requieren especial atención porque un cambio entre pesos y dólares puede generar diferencias aparentes muy grandes en `price_value` que no representan necesariamente una modificación equivalente del alquiler.

```python
# Cuenta cuántas monedas distintas aparecen para cada propiedad.
cambios_moneda = (
    df_clean
    .groupby("id")["currency"]
    .nunique()
)

print(
    "IDs con más de una moneda:",
    (cambios_moneda > 1).sum()
)
```

```python
# Muestra las propiedades que presentan cambios de moneda.
print(
    cambios_moneda[cambios_moneda > 1]
    .sort_values(ascending=False)
    .head(20)
)
```

```python
# Selecciona las propiedades con cambios de moneda para revisar su evolución.
ids_cambio_moneda = cambios_moneda[cambios_moneda > 1].index

cambios_moneda_detalle = (
    df_clean[df_clean["id"].isin(ids_cambio_moneda)]
    [["id", "Price", "price_value", "currency", "Location", "scrap_date"]]
    .sort_values(["id", "scrap_date"])
)
```

```python
# Muestra los cambios de moneda junto con la fecha del scraping.
print(
    cambios_moneda_detalle[
        ["id", "Price", "currency", "scrap_date"]
    ].to_string(index=False)
)
```

## Verificación de duplicados exactos

Se verifica si existen registros completamente idénticos en todas sus columnas.

Los registros de una misma propiedad en diferentes fechas no se consideran duplicados exactos, ya que pueden representar distintas versiones de una misma publicación.

La conversión temporal a texto permite realizar la comparación incluso cuando existen columnas que contienen listas.

```python
# Convierte temporalmente las columnas a texto y cuenta registros 
# completamente idénticos.

duplicados_exactos = df_clean.astype(str).duplicated().sum()

print(
    "Duplicados exactos:",
    duplicados_exactos
)
```

### Resultado

No se encontraron registros completamente duplicados en el dataset.

Aunque existen propiedades con el mismo `id` en diferentes fechas de scraping, estos registros no son duplicados exactos y pueden contener cambios en precio, moneda u otra información relevante.
